```python
#creating a function to sqaure a function
def y_squared(x):
    return x**2
y_squared(2)
```

```python
#squaring number in an array
x = np.array([1,2,3,4,5])
print(y_squared(x))
y=y_squared(x)
```

```python
#importing matplot to draw graph 
import matplotlib.pyplot as plt
```

```python
#plotting the graph
plt.plot(x,y,color='navy',linewidth=2)
plt.title('A graph of y against x')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```

```python
x = np.array([1,2,3,4,5])
y = x**3 + 1
plt.plot(x,y,color='black', label = 'y=x^3+1')
plt.title('A graph of y against x')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
```
