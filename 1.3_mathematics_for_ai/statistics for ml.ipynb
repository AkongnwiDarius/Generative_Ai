## Descriptive statistics
#### median is used when the value have outliers
#### mean is best for smaller values 
#### mode is used for text 

#### Mean

```python
#calculating the mean
prices = np.array([210,225,240,250,900])
print('mean: ',np.mean(prices))
print('mean(no outlier)',np.mean(prices[:-1]))
```

### Median

```python
#calculating the median
prices = np.array([210,225,240,250,900])
print('median: ',np.median(prices))

even_data = np.array([1,2,3,4])
print('median (even n): ',np.median(even_data))
```

### Mode

```python
#calculating the mode
shoe_sizes = np.array([8,9,9,10,9,11,8,9])
print(np.unique(shoe_sizes))
# mode_value = values[np.argmax(counts)]
# print('mode: ',mode_value, '! appears', counts.max(), 'times')
```

```python
print(np.unique(shoe_sizes, return_counts=True))
```

```python
print(np.argmax(np.array([2, 4, 1, 1]))) #gives the position of the largest occuring number in the array
values, frequency = np.unique(shoe_sizes,return_counts=True)
position_of_largest_frequency = np.argmax(frequency)
mode = values[position_of_largest_frequency]
print(mode)
```

### Range
the difference between the largest and the smallest

```python
temps = np.array([18,21,19,25,17,40])
print('range: ',np.ptp(temps))
print('range(no glitch): ',np.ptp(temps[:-1]))
```

## Variance and Standard Deviation

```python
scores = np.array([60,70,80,90,100])
mean = np.mean(scores)
deviations = scores - mean
sq_deviations = deviations ** 2
variance = np.var(scores, ddof=1)
std_dev = np.std(scores, ddof=1)
print('variance: ',variance)
print('std dev: ',std_dev.round(2))
```
