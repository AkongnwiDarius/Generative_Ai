Question 4

```python
query = np.array([3,4])
key = np.array([4,3])
a = np.array([])
total = sum(x*y for x,y in zip(query,key))
print(total)
print(query@key)
print(np.dot(query,key))
```

```python
fig, ax=plt.subplots(figsize=(6,4))
plt.quiver(0,0, query[0], query[1], angles='xy', scale_units='xy', scale=1, color='purple',label='query=[3,4]')
plt.quiver(0,0, key[0], key[1], angles='xy', scale_units='xy', scale=1, color='navy',label='key=[4,3]')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.legend()


ax.set_xlim(-1, 6)
ax.set_ylim(-1, 6)
```

```python
abs_q = np.linalg.norm(query)
abs_k = np.linalg.norm(key)
abs = abs_q*abs_k
print(abs)
```

```python
costheta = total/abs
print(costheta)
```

```python
theta = np.arccos(costheta)
theta
```

```python
angle = np.degrees(theta)
angle
```

```python
fig, ax=plt.subplots(figsize=(6,4))
plt.quiver(0,0, query[0], query[1], angles='xy', scale_units='xy', scale=1, color='purple',label='query=[3,4]')
plt.quiver(0,0, key[0], key[1], angles='xy', scale_units='xy', scale=1, color='navy',label='key=[4,3]')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.legend()


ax.set_xlim(-1, 6)
ax.set_ylim(-1, 6)
plt.text(2, 1, f'angle = {angle:.1f} degrees')
plt.show()
```
