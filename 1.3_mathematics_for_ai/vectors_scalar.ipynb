### length or magnitude of a vector

```python
v = np.array([3,4])
# method 1
manual = np.sqrt(v[0**2]+v[1]**2)
print(manual)
# method 2
print(np.linalg.norm(v)) #simply calculates the magnitude
print(np.linalg.norm(v, ord=1))  # takes the absolute magnitude of the vector and adds them
print(np.linalg.norm(v, ord=np.inf))  #compares the number in the vector and prints the highest

```

### Unit vector 
### This is a vector whose length is equals to one 
### it can be created by dividing a vector by its norm or magnitude which strips away the magnitude and keeps only direction

```python
v = np.array([3.0, 4.0])
unit_v = v/np.linalg.norm(v)
print(unit_v)
print(np.linalg.norm(unit_v)) # confirmed length is 1
```
