## Bag Of Words 

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# 1. Create a sample dataset with text reviews
data = {'reviews': ['the food is good',
                           'New food is terrible',
                           'good food and good services']
              }

df = pd.DataFrame(data)
df

# 2. Initialize the CountVectorizer
# stop_words='english' removes useless words like 'the' and 'and' to save memory
vectorizer = CountVectorizer(stop_words='english')

# 3. Fit and transform the text column 
# This learns the unique vocabulary and counts the word occurrences 
bow_matrix = vectorizer.fit_transform(df['reviews'])

# 4. Convert the matrix into a clean Pandas DataFrame
# vectorizer.get_feature_names_out() extracts the learned vocabulary as column names 

df_bow = pd.DataFrame(bow_matrix.toarray(), columns=vectorizer.get_feature_names_out())
```

### Tokenization

```python
#white space tokenization or white space splitting
text = 'Deep learning is amazing!'
tokens = text.split()

print(tokens)
```

```python
text = 'Deep learning is amazing!'
tokens = text.split('e')

print(tokens)
```

```python
#regex tokenization
import re 
text = 'Deep learning is amazing!'
tokens = re.findall(r'\w+|[^\w\s]', text)

print(tokens)
```

```python
# character level tokenization
text = 'Deep learning is amazing!'
tokens = list(text)

print(tokens)
```

### Step By Step implementation of a simple Tokenizers

```python
#opening the raw text file
with open('tokenizer.txt', 'r', encoding='utf-8') as f:
    raw_text = f.read()
len(raw_text),raw_text
```

```python
#whitespace splitting
import re
white_space_splitting = re.split(r'\s', raw_text)
print(white_space_splitting)
```

```python
len(white_space_splitting)
```

```python
#removing punctuations and stages
raw_text = re.sub(r'@\w+', '', raw_text, flags=re.MULTILINE)
raw_text = re.sub(r'#\w+', '', raw_text, flags=re.MULTILINE)

#removing anything that is not alphabet
raw_text = re.sub(r'[^a-zA-Z\s]', '', raw_text)

#removing any extra white spaces
raw_text = re.sub(r'\s+', ' ', raw_text).strip() 
print(raw_text)
```

```python
#splitting by other criteriars
white_space_splitting_with_special_xters = re.split(r'--|[@^*&.$%#@!??/]|\s', raw_text)
print(white_space_splitting_with_special_xters)
```

```python
#removes whitespaces from the text 
strip_text = [word.strip() for word in white_space_splitting_with_special_xters if word.strip()]
len(strip_text)
```

```python
print(strip_text)
```

```python
#removing duplicates and in alphabetical order
sorted_strip_text = sorted(set(strip_text))
print(sorted_strip_text)
len(sorted_strip_text)
```

```python
#enumerate keeps track of idex and words by pairing the index to the word 
for idx,token in enumerate(sorted_strip_text):
    print(f'( {idx}, {token})')
```

```python
#assigning the index to the words or tokens 
vocabulary = {}
for idx,token in enumerate(sorted_strip_text):
    vocabulary [token] = idx
vocabulary
```

```python
id_to_str = {}
for idx,token in enumerate(vocabulary):
    # print(f' ( {idx} , {token})')
    id_to_str[idx] = token
id_to_str
```

```python
#getting our tokenized text 
token_ids = [vocabulary[word] for word in strip_text]
print(token_ids)
```

```python
# recovering our tokenized text to its original state
tokens_recover = [id_to_str[idx] for idx in token_ids]
' '.join(tokens_recover) #.join converts a list to s string and .split converts a string to a list 
```
