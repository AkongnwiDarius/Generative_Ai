## string methods to know as an AI engineer



some string methods include .capitalize, .upper, .lower, .slice, indexing and accessing sting elements, .strip, .title, .endwith, .startwith

```python
text = '  generative, Ai, is, my, best field '
print(text.capitalize()) #captilazes the first word of the string
print(text.upper()) #capitalizes all the characters in the string
print(text.lower()) #converts all characters in a string to lower case
print(text.title()) # capitalizes the first character in each string
print(text.strip()) #remove spaces at the beginning and end of string characters but not in between
print(text.split(',')) #splits words at specified parameters given 
```

```python
print(len(text))
```

```python
remove_text = text
print(len(remove_text.strip()))
```

## Slicing

```python
greetings = 'Hello, SEED' #indexing begins from 0 and each character is counted be it a space or comma
greetings
print(greetings[0])
print(greetings[1])
print(greetings[2])
```

```python
greetings2 = 'MY GOAT IS MESSI'
# greetings2[0] + greetings2[1] + greetings2[2] + greetings[3] + greetings2[4] + greetings[5] + greetings[6] + greetings2[7]
print(greetings2[0:7])
print(greetings2[11:16]) #or 
print(greetings2[11:])    #which means print from the eleventh index to the last character

print(greetings2[3:7])







```
