## Conditionals and Loops

#### Control structure is the flow of the execution of a program or how a program is being executed
#### There are different types of control structure

#### 1. sequential which is simply running your code from top to bottom
#### 2. conditionals execute code if a particular condition is met 


```python
age = int(input('enter your age: ')) #converts the input to an interger because python automatically accepts the string data type when an input function is mention
if age > 20:
    print('age is greater than 20')

print('completed')
```
