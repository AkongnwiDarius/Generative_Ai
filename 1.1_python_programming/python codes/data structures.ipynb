```python
#takes user input names and store in a list and then checks foor duplicates and print it 
user_list = []
for name in range(1,6):
    user_name = input(f'enter your name: ')
    user_list.append(user_name)
print(f'the original names are: {user_list}')

duplicates = []
for names in user_list:
    if user_list.count(names) > 1 and names not in duplicates:
        duplicates.append(names)
if len(duplicates) > 0:
    print(duplicates)
else:
    print('non duplicate')
```

## Looping over a Dictionary
#### Looping by keys, by values and by items

```python
print(settings)
for key in settings.keys():
    print(key)
for value in settings.values():
    print(value)
for items in settings.items():
    print(items)
```

```python
dict = {
    'word': 'hello hello word'
}
for words in dict.values():
    print(words.count('hello'))
```

```python
sentence = 'God is the answer and they are saying that they are not tired'
sen = sentence.split()
print(sen)
s1 = ' '.join(sen)  #can reverse or join sentences back to its original
print(s1)
for worrds in sen:
    print(worrds)
print(sen)
```

```python
word_count = {}
for word in sen:
    if word in word_count.keys():
        word_count[word] += 1 
    else:
        word_count[word] = 1
word_count
```

### Sets and their methods

```python
#sets have no order
#no duplicates
#can be used to access elements faster

numbers = {1,2,3,4}
colors = {'red','green','blue'}
empty = set()
print(empty)
```

```python
#adding elements to a set
fruits = {'apple','banana'}
fruits.add('cherry')
print(fruits)

#updating elements
fruits.update(['orange','citrus'])
print(fruits)
```

```python
#removing elemets in a set
numbers_1 = {1,2,3,4}
numbers_1.remove(2) #throws an error if number or item is not found
numbers_1.discard(6) #doesnt t hrow an error 
print(numbers_1)

items = {'x','y','z'}
print(items.pop())  #removes and returs a random item hich has been removed because sets are unordered
```

```python
#union and intersection of sets
A = {1,2,3}
B = {3,4,5}
print(A | B)  #union
print(A & B)  #intersection
print(A - B)  #difference (items that are in A and not in B)
print(A ^ B)  #symmetric difference (items that are in one set or the other but not both)
```

```python
# looping in sets
numbers_list = {1,5,7,9}
for x in numbers_list:
    print(x)
```

```python
#set comprehension
sq = [i*i for i in range(1,11)]  #prints the first 10 square numbers
sq #also work for other data structures like list, dictionaried and tuples e.t.c., 
```
