```python
num1 = int(input('enter number 1: '))
num2 = int(input('enter number 2: '))
print (f'the sum of {num1} and {num2} is {num1 + num2}')
print (f'the difference of {num1} and {num2} is {num1 - num2}')
print (f'the product of {num1} and {num2} is {num1 * num2}')
print (f'the quotient of {num1} and {num2} is {num1 / num2:.0f}')  # rounds decimal places up or down 
print (f'the remainder of {num1} and {num2} is {num1 % num2}')
```

6. Assign grades as follows;
- 70-100 is 'A'
- 60-69 is 'B'
- 50-59 is 'C'
- 40-49 is 'D'
- Below 40 is 'f'

```python
user_score = int(input('enter your score: '))
if user_score >= 70 and user_score <= 100:
    print('Grade is A')
elif user_score >= 60 and user_score <= 69:
    print('Grade is B')
elif user_score >= 50 and user_score <= 59:
    print('Grade is C')
elif user_score >= 40 and user_score <= 49:
    print('Grade is D')
elif user_score <= 40 and user_score >=0:
    print('Grade is F')
else:
    print('unexpected value')
```

8. Ask the user for a word and count how many vowels it contains


```python
user_word = input('enter a word: ')
count_vowels= 0
vowels = ['a','e','i','o','u']
vowel_num = []
for word in user_word:
    if word in vowels:
        vowel_num.append(word)   #when it sees the vowel in adds it to the list created
        count_vowels = count_vowels +1    #now when it meets a vowel it counts it and then add to the count_vowels
print(vowel_num)
print(count_vowels)
```
