# Functions

```python
def greet(name, language='English'): #greet is the function name and name and language are the parameters it will take 
    '''say hello to someone in a chosen langauge'''
    if language == 'French':
        return f'Bonjour, {name}'
    return f'Hello, {name}'

print(greet('Alice'))
print(greet('Bob','French'))
```

```python
def greet1(name):
    print(f'hello {name}')
greet1('john')
```

```python
def greet2(name):
    return f'hello {name}'  #send the results back to the caller or signals for program ended
greet2('john')
```

```python
#creating a function to count elements in a list
def mylist(list):
    return f'{len(list)}'
list = [1,2,3,4,5,6]
print(mylist(list))

```

```python
def introduction(name,hobby, country):
    message = f'my name is {name} i a from {country} and i love {hobby}'
    return message
introduction('messi','football','cmr')
```

# *arg and **Kwargs

```python
def total(*values):   # *args: this parameter takes in many positinal argument elements and stores in a tuple
    t=sum(values)     # **kwargs: this parameter takes any number of keyword arguments(in pairs) and stores in a dictionary
    return t
total(1,2,3,4,5,6,7,8,9)
```

```python
def computer_spec(**spec):
    return spec
computer_spec(color = 'pink',RAM = '16GB')
```

```python
def computer_spec2(**spec2):
    for i in spec2.items():
        print(i)
    return spec2
computer_spec2(x=2,y=16,z=40)
```

# Lamda Functions

```python
square = lambda X: X**2
square(5)
```

```python
#sorting using lambda
laliga_table = [('ATM,15'),('Bar,22'),('RMA,19')]
laliga_table.sort(key=lambda m: m[2], reverse = True)
laliga_table
```

```python
greet = lambda name: f'hello {name}'
greet('john')
```
